# Set up data collection

```python
def extract_urls_from_url(url):
    # Create a cloudscraper instance to handle the request bypass
    scraper = cloudscraper.create_scraper()
    response = scraper.get(url)

    if response.status_code == 200:
        html_content = response.text
        soup = BeautifulSoup(html_content, 'html.parser')

        # Replace 'related_aesthetics' with the actual data source value you're looking for
        divs_with_data = soup.find_all('div', {'data-source': 'related_aesthetics'})

        if divs_with_data:
            # List to store extracted URLs
            url_list = []

            # Extract URLs from each div
            for div_with_data in divs_with_data:
                # Find all <a> tags within the div
                links = div_with_data.find_all('a', href=True)

                # Extract and append the href attribute to the list
                for link in links:
                    url_list.append(link['href'])

            return [l[6:] for l in url_list]
        else:
            print(f"{url[35:]}: Divs with data source 'related_aesthetics' not found.")
            return None

    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")
        return None

def get_colors(n):
    return ["#%06x" % random.randint(0, 0xFFFFFF) for _ in range(n)]

def extract_data_from_url(url):
    # Create a cloudscraper instance to handle the request bypass
    scraper = cloudscraper.create_scraper()
    response = scraper.get(url)

    if response.status_code == 200:
        html_content = response.text
        soup = BeautifulSoup(html_content, 'html.parser')

        # Data sources you're interested in
        data_sources = ['related_aesthetics', 'decade_of_origin', 'key_motifs', 'key_colours', 'key_values', 'related_media']

        # Dictionary to store extracted data
        extracted_data = {}
        
        no_get = ''
        
        for source in data_sources:
            divs_with_data = soup.find_all('div', {'data-source': source})

            if divs_with_data:
                # Lists to store extracted data for the current source
                urls = []
                text_data = []

                # Extract URLs or text from each div
                for div_with_data in divs_with_data:
                    if source == 'related_aesthetics':
                        # Extract URLs
                        links = div_with_data.find_all('a', href=True)
                        urls.extend(link['href'] for link in links)
                    else:
                        # Extract text
                        text = div_with_data.get_text(strip=True)
                        text_data.append(text)

                # Add the extracted data to the dictionary
                extracted_data[source] = (urls, text_data)

            else:
                no_get += f', {source}'
        if len(no_get) > 0:
            print(f"      {url[35:]}: {no_get[1:]} not found.")
        
        try:
            extracted_data['related_aesthetics'] = [l[6:] for l in extracted_data['related_aesthetics'][0]]
        except:
            extracted_data['related_aesthetics'] = []
        
        for d in ['decade_of_origin', 'key_motifs', 'key_colours', 'key_values', 'related_media']:
            try:
                extracted_data[d] = extracted_data[d][1][0][len(d):]
            except:
                extracted_data[d] = ""
        
        return extracted_data

    else:
        print(f"Failed to retrieve the page. Status code: {response.status_code}")
        return None
```

```python
# Function to clean the column and extract the smallest number, converting shorthand decades
def clean_to_number(text):
    # Regular expression to find all numbers
    numbers = re.findall(r'\d+', text)
    
    # Check for shorthand decade formats like '90s' and convert to full year format
    if re.search(r'\b\d{1,2}s\b', text):
        # Extract the number part (like 90 in '90s')
        shorthand_decade = re.search(r'\d{1,2}', text).group()
        # Convert to full decade (e.g., 90 -> 1990, 80 -> 1980)
        full_decade = int(shorthand_decade) + (1900 if int(shorthand_decade) < 100 else 0)
        return full_decade
    
    if numbers:
        # Convert found numbers to integers and return the smallest one
        return min(map(int, numbers))
    else:
        # If no numbers found, return None
        return None
        
# Convert all entries in 'decade_of_origin' to strings
table['decade_of_origin'] = table['decade_of_origin'].astype(str)
# Apply the function to clean the 'decade_of_origin' column
table['earliest_origin'] = table['decade_of_origin'].apply(clean_to_number)

key_adj = ['Cottagecore', 'Aestheticism', 'Afropunk', 'American_Gothic', 'Ancient_Egypt', 'Arabian_Nights', 'Art_Deco', 'Australian_Outback', 'Baroque', 'Burlesque', 'Cassette_Futurism', 'Classicism', 'Clockpunk', 'Dionysism', 'Driftcore', 'Expressionism', 'Fanfare', 'Flamenco', 'French_Provincial_Style', 'Gangsta_Rap', 'Goofcore', 'Gothic', 'Hair_Metal', 'Italian_Renaissance', 'Kiwiana', 'Maximalism', 'Medieval', 'Modernism', 'Mud%C3%A9jar', 'Neo_Baroque', 'Post-Apocalyptic', 'Rangercore', 'Rebel_Academia', 'Rustic', 'Saint_Patrick%27s_Day', 'Scrapbook', 'Southern_Gothic', 'Spanish_Golden_Age', 'Spy_Fiction', 'Steampunk', 'Teslapunk', 'Tranquil_Spa_Graphics', 'Tuscan_Rustic', 'Vampire', 'Viking', 'Werewolf']
year_adj = [2010, 1900, 1980, 1850, -30, 1850, 1910, 0, 1750, 1700, 1970, -500, 1500, -500, 1995, 1900, 1800, 1400,1600, 1980, 1980, 1100, 1980, 1650, 1950, 1850, 400, 1900, 1500, 1980, 1800, 2019, 1950, 1800, 900, 1500, 1800, 1600, 1900, 1980, 2000, 400, 1975, 1800, 700, 1910]

for i in range(len(key_adj)):
    table.loc[table['key'] == key_adj[i],'earliest_origin'] = year_adj[i]

table.to_csv('Aesthetic network.csv', index=False)
table
```

```python
# Helper function to find and parse ALL years/decades inside the text
def parse_all_years(text):
    text = str(text).strip().lower()
    if not text or text == 'nan':
        return []
    
    found_years = []

    # Standardize all dashes right away (including box lines like ─)
    text = re.sub(r'[–—−─]', '-', text)

    # 1. Handle Century Ranges and Single Centuries (e.g., "14th-16th centuries", "19th century")
    century_matches = re.findall(r'(\d{1,2})(?:st|nd|rd|th)?(?:\s*-\s*(\d{1,2})(?:st|nd|rd|th)?)?\s+centur', text)
    if century_matches:
        for c1, c2 in century_matches:
            if c1:
                v1 = int(c1)
                found_years.extend([(v1 - 1) * 100, v1 * 100])
            if c2:
                v2 = int(c2)
                found_years.extend([(v2 - 1) * 100, v2 * 100])
        # FIX: Erase the century wording from the text so it doesn't interfere with standard number parsing
        text = re.sub(r'\d{1,2}(?:st|nd|rd|th)?(?:\s*-\s*\d{1,2}(?:st|nd|rd|th)?)?\s+centur\w*', '', text)

    # 2. Find shorthand chunks (e.g., 90's, 90s, 00's, 2000's)
    shorthands = re.findall(r'(\d{1,4})\'?s', text)
    for item in shorthands:
        val = int(item)
        if val < 100:  
            # Cut-off rule: < 20 means 2000s, >= 20 means 1900s
            full_decade = val + (2000 if val < 20 else 1900)
            found_years.append(full_decade)
        else:  
            found_years.append(val)
            
    # 3. Find standalone numbers
    # Strip away the shorthands we already evaluated to prevent double-counting
    clean_text = re.sub(r'\d{1,4}\'?s', '', text)
    
    # Check if the text explicitly states BC or BCE
    is_bc = 'bc' in clean_text or 'bce' in clean_text

    standalone_numbers = re.findall(r'\d+', clean_text)
    for item in standalone_numbers:
        val = int(item)
        # Only turn it negative if it's explicitly a BC/BCE date
        if is_bc:
            val = -abs(val)
        found_years.append(val)
        
    return found_years

# Function to extract the smallest calculated year
def clean_to_earliest(text):
    years = parse_all_years(text)
    return min(years) if years else None

# Function to extract the largest calculated year
def clean_to_latest(text):
    years = parse_all_years(text)
    return max(years) if years else None

# Convert all entries in 'decade_of_origin' to strings
table['decade_of_origin'] = table['decade_of_origin'].astype(str)

# Apply the functions to clean the columns
table['earliest_origin'] = table['decade_of_origin'].apply(clean_to_earliest)
table['latest_origin'] = table['decade_of_origin'].apply(clean_to_latest)

# Hardcoded adjustments
manual_adjustments = {
    'Cottagecore': 2010,                        # "2010s (inspired by the 19th century)" -> Modern internet era
    'Aestheticism': (1860, 1900),               # "Late-19th century (approximately 1860s-1900s)"
    'Afropunk': (1970, 2000),                   # "1970s (origins)2000s (wider popularization)"
    'American_Gothic': 1850,                    # Flatten historic baseline
    'Ancient_Egypt': (-3150, -30),               # "3150 BC─30 BC" -> True chronological bounds
    'Arabian_Nights': 2020,                     # Modern interpretation decade tracking
    'Art_Deco': (1910, 1930),                   # "1910s–1930s"
    'Australian_Outback': 1900,                  # Approximate historical tracking match
    'Baroque': (1600, 1750),                    # "c. 1600 - 1750"
    'Burlesque': (1600, 1700),                  # "17th Century"
    'Cassette_Futurism': 1970,                  # "1970s"
    'Classicism': (-500, -400),                 # "5th century BCE (Original Period)" -> Correctly negative
    'Clockpunk': (1300, 1600),                  # "Renaissance inspiration (14th–17th centuries)" -> century start bounds
    'Dionysism': (-1200, -600),                 # "13th-7th centuries BC" -> Correct BC ranges
    'Driftcore': 1995,                          # Fixed alignment
    'Expressionism': (1900, 1920),              # "1900s-1920s"
    'Fanfare': (1742, 1900),                    # "Created 1742, popularized late 19th century"
    'Flamenco': (1700, 1800),                   # "Late 18th century (modern-day form)"
    'French_Provincial_Style': (1600, 1800),    # "17th-18th Century"
    'Gangsta_Rap': 1980,                        # Standard 80s baseline 
    'Goofcore': 1980,                           # 80s movement baseline
    'Gothic': (1700, 1800),                     # "Late 18th century (c. 1764)"
    'Hair_Metal': (1970, 1980),                 # "Late 70's-early 80's"
    'Italian_Renaissance': (1300, 1600),        # "14th-16th Centuries"
    'Kiwiana': (1900, 2000),                    # "Around the middle of the 20th century"
    'Maximalism': (1800, 1900),                 # "19th Century"
    'Medieval': (400, 500),                     # "5th Century AD"
    'Modernism': 1900,                          # Standard 20th-century baseline
    'Mud%C3%A9jar': (1100, 1600),               # "12th–16th Centuries"
    'Neo_Baroque': 1980,                        # Modern styling revival
    'Post-Apocalyptic': (1800, 1900),           # "Biblical stories, 19th Century works"
    'Rangercore': 2019,                         # Internet subculture arrival
    'Rebel_Academia': 1950,                     # Mid-century stylistic backdrop
    'Rustic': 1800,                             # Fallback for completely empty/NaN row
    'Saint_Patrick%27s_Day': 900,               # Legacy cultural mapping
    'Scrapbook': 1500,                          # Historical artifact alignment
    'Southern_Gothic': (1800, 1900),            # "19th Century"
    'Spanish_Golden_Age': (1492, 1681),         # "c. 1492 - 1681"
    'Spy_Fiction': (1800, 1900),                # "19th century"
    'Steampunk': (1960, 2010),                  # "1960s (aesthetic)1980s (term)2000s/2010s"
    'Teslapunk': 2000,                          # "2000s19th century (concept)" -> Modern aesthetic timeline
    'Tranquil_Spa_Graphics': 2000,              # Y2K corporate aesthetic grouping
    'Tuscan_Rustic': 2000,                      # "Traditional, popularized in the mid-2000s"
    'Vampire': (1800, 1900),                    # "19th Century (Modern Archetype)"
    'Viking': 700,                              # Historic baseline bounds
    'Werewolf': (1910, 1910)                    # Skip ancient legend text noise to target modern aesthetic trope
}

for key, adj_val in manual_adjustments.items():
    if isinstance(adj_val, tuple):
        table.loc[table['key'] == key, 'earliest_origin'] = adj_val[0]
        table.loc[table['key'] == key, 'latest_origin'] = adj_val[1]
    else:
        table.loc[table['key'] == key, 'earliest_origin'] = adj_val
        table.loc[table['key'] == key, 'latest_origin'] = adj_val

table.to_csv('Aesthetic network.csv', index=False)
table
```
